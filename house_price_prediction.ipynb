```python
import pandas as pd

#reads train.csv and displays 5 first rows
train = pd.read_csv('train.csv')
print (train.head())
```

```python
# does the same with test.csv
test = pd.read_csv('test.csv')
print (test.head())

#output has 80 columns cause we want the ML to predict the saleprice column
```

```python
#filters numeric values with data type float and int
numeric_values = train.select_dtypes(include=[float, int])

#get correlation value of numeric columns
correlation_values= numeric_values.corr()

# Sorting the correlation values with respect to SalePrice in descending order
numeric_cols_saleprice_corr = correlation_values["SalePrice"].sort_values(ascending=False)

# Display the top correlations with SalePrice
numeric_cols_saleprice_corr.head(7)
```

```python
# gets the "sum" when missing values (true) divides it by the dataframe lenght then x100 to get it in a form %
missing_count = (train.isnull().sum() / len(train))*100

# Displays the missing count greater than 0
missing_count = missing_count[missing_count > 0]
print(missing_count.sort_values())
```

```python
# Drop columns with more than 50% missing values.
train = train.dropna(thresh=0.5*len(train), axis=1)
print (train.columns)
print (train.shape)

#columns : 81 - 5 =76
```

```python
#displays the number of numeric & categorical values.
numeric_columns = train.select_dtypes(include=[float, int]).columns

# filters categorical columns, columns whihc are objects
categorical_columns = train.select_dtypes(include=['object', 'category']).columns.tolist()

# Count the number of numeric and categorical columns
total_numeric_columns = len(numeric_columns)
total_categorical_columns = len(categorical_columns)

# Print the counts
print(f"Number of numeric columns: {total_numeric_columns}")
print(f"Number of categorical columns: {total_categorical_columns}")

# columns : 38 + 38 = 76
```

```python
#determine and handle missing numeric columns :

# Filter for columns with missing values
numeric_cols_with_missing = train[numeric_columns].columns[train[numeric_columns].isnull().any()]

# Fill missing values in numeric columns with the mean
for col in numeric_cols_with_missing:
    train[col] = train[col].fillna(train[col].mean())

# Display only numeric columns with missing values .. whihc is filled
print(train[numeric_cols_with_missing])
```

```python
#determine and handle missing categorical columns:

# Filter for columns with missing values
categorical_cols_with_missing = train[categorical_columns].columns[train[categorical_columns].isnull().any()]

# Fill missing values in categorical columns with the mode
for col in categorical_cols_with_missing:
    train[col] = train[col].fillna(train[col].mode()[0])

# Display only categorical columns with missing values .. which is filled
print(train[categorical_cols_with_missing])
```

```python
#OrdinalEncoding, to convert categorical data to nuemerical data
from sklearn.preprocessing import OrdinalEncoder

#common categorical features between train and test dataset
categorical_features = list(set(categorical_columns) & set(test.columns))

#auto fit the OrdinalEncoder using common features (selected columns)
OEncoder = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
OEncoder.fit(train[categorical_features])

#Transform the data for both train and test using common features
train[categorical_features] = OEncoder.transform(train[categorical_features])
test[categorical_features] = OEncoder.transform(test[categorical_features])

#drop 5 columns with above 50% missing values .. to avoid error
test= test.dropna(thresh=0.5*len(test), axis=1)

print(train)
```

```python
#split data into train & test sets
from sklearn.model_selection import train_test_split

# Separate features (other columns) and target (SalePrice)
feature = train.drop('SalePrice', axis=1)
target = train['SalePrice']

# slit : 1 - 0.2 = 0.8 => 80%train , 20%test
feature_train, feature_test, target_train, target_test = train_test_split(feature, target, test_size=0.2, random_state=42)

#print shape for each set
print("Training set shape:", feature_train.shape, target_train.shape)
print("Testing set shape:", feature_test.shape, target_test.shape)
```

```python
# Prediction:
from sklearn.ensemble import RandomForestRegressor

# create  random Forest Regressor model object
RF = RandomForestRegressor(n_estimators=100, random_state=42)

# Train the model
RF.fit(feature_train, target_train)

# Predict saleprice in test
prediction = RF.predict(test)

# output format id .. saleprice
output = pd.DataFrame({'Id': test['Id'], 'SalePrice': prediction})
print(output) #prints first 5 and last 5 output
#print(output.to_string()) #prints whole thing
#output.to_excel('output.xlsx', index=False) #print to excel file
```
